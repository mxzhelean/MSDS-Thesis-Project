### Execution: N=15

__Raw JSONL integrity check (verify 90/90 complete records)__

Check counts, status, and word distribution across models/tasks, plus 1 random sample


```python
from src.schema import load_records_jsonl

BATCH_NAME = "pilot_n15"
raw_files = sorted((Path("data/raw") / BATCH_NAME).glob("*.jsonl"))

records = []
for f in raw_files:
    records.extend([r.to_dict() for r in load_records_jsonl(f)])

raw_df = pd.DataFrame(records)
print(f"Total loaded records: {len(raw_df)}")

# Check success rates across models and tasks
display(pd.crosstab([raw_df["model_id"], raw_df["task_id"]], raw_df["status"]))

# Sanity check: authentic raw output sample for Task 2 (AI Ethics)
sample_t2 = raw_df[raw_df["task_id"] == "Task2_AIEthics"].iloc[0]
print(f"\n--- Preview: {sample_t2['sample_id']} ({sample_t2['word_count']} words) ---")
print(sample_t2["raw_output"][:500] + "\n...")
```

__Run automated quantitative analysis (CV, TTR/MTLD, and semantic cosine similarity)__

Invoke run_quantitative_analysis() to compute all metrics and display summary statistics in an interactive table


```python
from src.analytics.quant_metrics import run_quantitative_analysis

sample_df, summary_df, pairwise_df = run_quantitative_analysis(
    batch_name=BATCH_NAME,
    embed_model_name="all-MiniLM-L6-v2",
)

# summary table
display_cols = [
    "model_label",
    "task_id",
    "n_samples",
    "wc_mean",
    "wc_sd",
    "wc_cv",
    "window_hit_rate",
    "ttr_mean",
    "mtld_mean",
    "pairwise_cos_mean",
    "pairwise_cos_sd",
]
display(
    summary_df[display_cols].style.background_gradient(
        subset=["wc_cv", "mtld_mean", "pairwise_cos_mean"], cmap="YlGnBu"
    )
)
```

```python
from src.analytics.plot_metrics import PALETTE, configure_style

configure_style()
model_order = [m for m in PALETTE if m in sample_df["model_label"].unique()]

fig, axes = plt.subplots(1, 3, figsize=(18, 5.2))

# Subplot 1: Word count distribution and 400–450 word target band
axes[0].axhspan(400, 450, color="#2ECC71", alpha=0.15, label="Target (400–450w)")
sns.boxplot(
    data=sample_df,
    x="task_id",
    y="word_count",
    hue="model_label",
    hue_order=model_order,
    palette=PALETTE,
    width=0.55,
    ax=axes[0],
)
axes[0].set_title("1. Output Length & Target Compliance", fontweight="bold")
axes[0].set_xlabel("")
axes[0].set_ylabel("Word Count")
axes[0].get_legend().remove()

# Subplot 2: Length-independent lexical diversity (MTLD)
sns.boxplot(
    data=sample_df,
    x="task_id",
    y="mtld",
    hue="model_label",
    hue_order=model_order,
    palette=PALETTE,
    width=0.55,
    ax=axes[1],
)
axes[1].set_title("2. Lexical Diversity (MTLD)", fontweight="bold")
axes[1].set_xlabel("")
axes[1].set_ylabel("MTLD Score")
axes[1].get_legend().remove()

# Subplot 3: Inter-run semantic similarity (105 pairwise cosine scores)
sns.violinplot(
    data=pairwise_df,
    x="task_id",
    y="cosine_similarity",
    hue="model_label",
    hue_order=model_order,
    palette=PALETTE,
    inner="box",
    cut=0,
    ax=axes[2],
)
axes[2].set_title("3. Cross-Iteration Semantic Homogenization", fontweight="bold")
axes[2].set_xlabel("")
axes[2].set_ylabel("Pairwise Cosine Similarity")
axes[2].legend(title="Model", loc="lower right")

plt.tight_layout()
plt.show()
```
